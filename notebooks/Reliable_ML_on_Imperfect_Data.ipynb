# **Reliable Machine Learning on Imperfect Data**

### A Study of Model Robustness, Prediction Reliability, and Data Quality

---

## Research Overview

Machine learning models are commonly evaluated on datasets that have
already undergone substantial preprocessing and cleaning. However,
real-world datasets frequently contain imperfections such as missing
values, outliers, and incorrect labels.

These imperfections can affect both predictive performance and the
reliability of machine learning predictions.

This project investigates how different types and levels of data
imperfections affect machine learning models. It further examines
whether model disagreement and prediction confidence can help identify
samples for which model predictions may be less reliable.

The study follows a controlled experimental approach in which data
imperfections are systematically introduced and their effects are
measured against a clean-data baseline.

---

## Research Objective

The primary objective of this study is to systematically evaluate the
robustness and prediction reliability of machine learning models when
trained on imperfect data.

The study focuses on four major aspects:

1. The effect of missing data on model performance.
2. The effect of feature outliers on model performance.
3. The effect of label noise on model performance.
4. The use of model disagreement and prediction confidence to identify
   potentially unreliable samples.

The study also investigates whether appropriate data-quality
interventions can recover performance after data corruption.

---

## Research Questions

### RQ1 — Missing Data

How does increasing missingness affect the predictive performance and
reliability of different machine learning models?

### RQ2 — Outliers

How sensitive are different machine learning models to increasing
levels of feature outliers?

### RQ3 — Label Noise

How does increasing label noise affect model performance and prediction
reliability?

### RQ4 — Model Disagreement

Do different machine learning models disagree more frequently on
samples that are difficult to classify or affected by data
imperfections?

### RQ5 — Unreliable Samples

Can model disagreement and prediction confidence be used to identify
potentially unreliable or difficult-to-classify samples?

### RQ6 — Recovery

Can appropriate data-quality interventions recover model performance
after data corruption?

---

## Research Hypotheses

### H1 — Missing Data

Increasing levels of missing data will generally reduce predictive
performance, although the magnitude of degradation will vary across
different machine learning models.

### H2 — Outliers

Increasing levels of feature outliers will negatively affect some
machine learning models more strongly than others.

### H3 — Label Noise

Increasing label noise will reduce model performance and make learned
predictions less reliable.

### H4 — Model Disagreement

Model disagreement will be higher for samples that are difficult to
classify or affected by data imperfections.

### H5 — Unreliable Samples

Samples associated with high model disagreement and low prediction
confidence are more likely to represent unreliable or difficult-to-
classify cases.

### H6 — Recovery

Appropriate data-quality handling can partially recover performance
lost due to data imperfections.

---

## Experimental Principle

The study will use a clean-data baseline and controlled levels of data
corruption.

For each type of imperfection, the corrupted datasets will be evaluated
using consistent data splits, preprocessing procedures, machine learning
models, and evaluation metrics.

The observed results will then be compared with the clean-data baseline.

> **Research workflow:**  
> Question → Hypothesis → Experiment → Result → Interpretation → Conclusion

# Experimental Design

The objective of this study is to evaluate the reliability and robustness of machine-learning models when the quality of the input training data is systematically degraded.

Instead of modifying the dataset arbitrarily, controlled data imperfections will be introduced at predefined severity levels. Model performance under each corrupted condition will then be compared with the performance obtained on the original clean dataset.

This experimental design allows the effect of individual data-quality problems to be isolated and measured.

### Experimental Reference

The original dataset will serve as the **clean reference condition**.

All corruption experiments will start from the same clean dataset and will use fixed random seeds to improve reproducibility.

### Data Imperfections Studied

The study will investigate three major forms of data imperfection:

1. **Missing Data**
   - Controlled missing values will be introduced into selected features.
   - Multiple missingness levels will be evaluated.
   - Appropriate imputation methods will then be investigated for recovery.

2. **Feature Outliers**
   - Controlled extreme values will be introduced into numerical features.
   - Different contamination levels will be evaluated.
   - Model degradation will be compared across models.

3. **Label Noise**
   - A controlled proportion of training labels will be randomly flipped.
   - Multiple noise levels will be evaluated.
   - Models will always be evaluated against the original clean test labels.

### Model Comparison

Multiple machine-learning models with different learning characteristics will be compared.

This is important because different algorithms may respond differently to the same type of data imperfection.

The models will therefore be evaluated not only according to their clean-data performance but also according to how much their performance changes as data quality deteriorates.

### Evaluation Strategy

For every experiment, the following quantities will be recorded:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

Performance degradation will be measured relative to the corresponding clean baseline.

### Reliability Analysis

Beyond predictive performance, the study will investigate whether model disagreement and prediction confidence can provide useful signals for identifying potentially unreliable or difficult-to-classify observations.

The analysis will examine whether observations associated with:

- high disagreement between models,
- low prediction confidence, or
- both conditions simultaneously

are more likely to correspond to prediction errors.

### Recovery Analysis

After measuring performance degradation, suitable data-quality interventions will be applied to corrupted training data.

The recovered models will then be compared with:

1. the clean baseline,
2. the corrupted-data model, and
3. the recovered model.

This will allow the study to quantify whether appropriate data-quality handling can recover performance lost due to data imperfections.

### Core Research Principle

Each experiment will follow the sequence:

**Research Question → Hypothesis → Controlled Experiment → Quantitative Results → Interpretation → Conclusion**

No experiment will be performed solely to maximize predictive performance. The primary objective is to understand how data imperfections affect model behavior and prediction reliability.

## Experimental Severity Levels

To study how model reliability changes as data quality deteriorates, each type of imperfection will be introduced at multiple severity levels.

Using multiple levels is important because a single corruption level would only show whether a model is affected. Multiple levels allow us to study the **relationship between corruption severity and model degradation**.

The initial experimental levels are defined as follows:

| Data Imperfection | Severity Levels |
|---|---|
| Missing Data | 5%, 10%, 20%, 30% |
| Feature Outliers | 1%, 5%, 10% |
| Label Noise | 5%, 10%, 20%, 30% |

The clean dataset represents the **0% corruption baseline**.

These levels will be applied consistently across models wherever applicable.

The exact corruption mechanism will be defined separately before each experiment so that the experiment remains reproducible and interpretable.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
```

## Baseline Model Selection

Three machine-learning models are selected for the clean baseline experiment:

1. Logistic Regression
2. Random Forest
3. Gradient Boosting

The models are intentionally chosen from different learning approaches so that the later robustness experiments can examine whether model behavior under data imperfections depends on the underlying learning mechanism.

### 1. Logistic Regression

Logistic Regression is used as a linear and relatively interpretable baseline model.

It estimates the probability of the binary target using a linear combination of the input features.

It provides a useful reference point because its behavior is relatively simple compared with tree-based ensemble models.

**Role in this study:**
- Linear baseline
- Interpretable model behavior
- Sensitive to feature scaling
- Useful for comparing linear and non-linear approaches

### 2. Random Forest

Random Forest is an ensemble of decision trees trained using the bagging principle.

Multiple decision trees are trained using different samples and feature subsets, and their predictions are combined to produce the final prediction.

Random Forest is capable of modeling non-linear relationships and interactions between features without requiring explicit specification of those relationships.

**Role in this study:**
- Non-linear tree-based model
- Bagging-based ensemble
- Less dependent on feature scaling
- Provides a contrast to the linear Logistic Regression model

### 3. Gradient Boosting

Gradient Boosting is a sequential ensemble learning method.

Instead of building trees independently, it builds a sequence of decision trees where later trees focus on improving the errors made by the existing ensemble.

This provides a different ensemble learning mechanism from Random Forest.

**Role in this study:**
- Non-linear tree-based model
- Boosting-based ensemble
- Sequential error-correction approach
- Provides a second type of ensemble model for robustness comparison

### Why These Three Models?

The three models provide methodological diversity:

| Model | Learning Approach | Main Role |
|---|---|---|
| Logistic Regression | Linear modeling | Interpretable baseline |
| Random Forest | Bagging | Non-linear ensemble |
| Gradient Boosting | Sequential boosting | Non-linear boosting ensemble |

This diversity is important for the research objective. The study is not only interested in overall predictive performance, but also in whether different learning approaches respond differently when the training data contains missing values, outliers, or label noise.

Therefore, the clean baseline will establish both the predictive performance and the initial behavioral differences between these three models.

```python
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", logistic_preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ]
)
```

## Clean Baseline Evaluation

The three selected models are now evaluated on the clean dataset before any artificial data imperfections are introduced.

A common evaluation framework is used so that the models can be compared under the same experimental conditions.

The following metrics are reported:

- **Accuracy:** Overall proportion of correct predictions.
- **Precision:** Proportion of predicted default cases that are actually defaults.
- **Recall:** Proportion of actual default cases correctly identified.
- **F1-score:** Harmonic mean of precision and recall.
- **ROC-AUC:** Measures the model's ability to distinguish between the two classes across different classification thresholds.

Because the target variable is moderately imbalanced, no single metric is treated as sufficient for evaluating the models.

The clean baseline results will serve as the reference point for all subsequent data-imperfection experiments.

For each corrupted dataset, the same evaluation procedure will be used and the resulting performance will be compared with this clean baseline.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

def evaluate_model(model, X_train, y_train, X_eval, y_eval):
    model.fit(X_train, y_train)

    y_pred = model.predict(X_eval)
    y_prob = model.predict_proba(X_eval)[:, 1]

    results = {
        "Accuracy": accuracy_score(y_eval, y_pred),
        "Precision": precision_score(y_eval, y_pred),
        "Recall": recall_score(y_eval, y_pred),
        "F1": f1_score(y_eval, y_pred),
        "ROC-AUC": roc_auc_score(y_eval, y_prob)
    }

    return results
```

```python
models = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model,
    "Gradient Boosting": gradient_boosting_model
}

baseline_results = {}

for name, model in models.items():
    baseline_results[name] = evaluate_model(
        model,
        X_train,
        y_train,
        X_val,
        y_val
    )

baseline_results
```

## Experiment 1: Robustness to Missing Data

### Research Question

**RQ1:** How does increasing missingness in the training data affect the
predictive performance of different machine learning models?

### Objective

Real-world datasets often contain missing values due to incomplete data
collection, measurement failures, or unavailable information. A model trained
on increasingly incomplete data may experience a degradation in predictive
performance.

In this experiment, we systematically introduce different levels of missing
values into the training features and study how the performance of the
selected machine learning models changes.

### Experimental Setup

The following missingness levels will be investigated:

- 0% — Clean training data
- 5% — Low missingness
- 10% — Moderate missingness
- 20% — High missingness
- 30% — Severe missingness

Missing values will be introduced **only into the training features**.

The validation set will remain completely clean and will be used to evaluate
the models. This allows us to measure how degradation in training-data quality
affects performance on previously unseen, clean data.

### Experimental Principle

For each missingness level:

1. Start with the original training data.
2. Introduce missing values at the specified rate.
3. Apply the required preprocessing and imputation using the training data.
4. Train the frozen model configurations selected during the baseline stage.
5. Evaluate the models on the clean validation set.
6. Record Accuracy, Precision, Recall, F1-score, and ROC-AUC.

The same random seed and corruption procedure will be used across experiments
to ensure reproducibility.

### Hypothesis

**H1:** Increasing levels of missing data will generally reduce predictive
performance, although the magnitude of degradation will differ across machine
learning models.

The experiment will also help determine whether some models are more robust
to missing training information than others.

### Missingness Generation Method

Missing values are introduced at the feature-cell level rather than by removing
entire observations.

For a given missingness level \(p\), approximately \(p\%\) of all feature cells
in the training set are randomly replaced with `NaN`.

Only the training feature matrix is corrupted:

- `X_train` → corrupted
- `y_train` → unchanged
- `X_val` → unchanged
- `y_val` → unchanged

The validation data remains clean so that model performance is evaluated under
a consistent reference condition.

A fixed random seed is used to make the corruption process reproducible.

The missingness mechanism used in this experiment is **Missing Completely At
Random (MCAR)**. Each feature cell has an equal probability of being selected
for corruption, independent of its feature value or target class.

The following missingness levels will be tested:

**0%, 5%, 10%, 20%, and 30%.**

### Verifying the Missingness Mechanism

Before running the complete missing-data experiment, the corruption procedure is
first tested at the 5% missingness level.

This verification step checks whether:

- approximately/exactly 5% of the training feature cells are missing,
- the target variable remains unchanged,
- the validation dataset remains completely clean, and
- the corruption procedure is reproducible.

Only after these checks pass will the same procedure be applied to the remaining
missingness levels of 10%, 20%, and 30%.

```python
def introduce_mcar_missingness(X, missing_rate, random_state=42):
    """
    Introduce MCAR missing values into a copy of X.

    missing_rate: proportion of feature cells to replace with NaN.
    """
    X_corrupted = X.copy().astype(float)

    rng = np.random.default_rng(random_state)

    n_cells = X_corrupted.shape[0] * X_corrupted.shape[1]
    n_missing = int(round(missing_rate * n_cells))

    missing_indices = rng.choice(
        n_cells,
        size=n_missing,
        replace=False
    )

    X_corrupted.values.flat[missing_indices] = np.nan

    return X_corrupted
```

## Imputation Strategy

After introducing MCAR missingness, missing values must be handled before
training the machine learning models.

The imputation strategy is selected according to the semantic type of each
feature rather than applying a single method to all columns.

### Numerical Features

The following features are treated as numerical:

- `LIMIT_BAL`
- `AGE`
- `BILL_AMT1` to `BILL_AMT6`
- `PAY_AMT1` to `PAY_AMT6`

**Imputation method: Median**

The median is used because several financial variables in the dataset are
skewed and contain large values. Compared with the mean, the median is less
sensitive to extreme observations.

### Nominal Categorical Features

The following features are treated as nominal categorical variables:

- `SEX`
- `MARRIAGE`

**Imputation method: Most Frequent**

These variables represent categorical groups, for which numerical statistics
such as mean or median are not meaningful. Missing values are therefore
replaced with the most frequently observed category in the training data.

### Ordinal Features

The following variables contain ordered/discrete codes:

- `EDUCATION`
- `PAY_0`
- `PAY_2`
- `PAY_3`
- `PAY_4`
- `PAY_5`
- `PAY_6`

**Imputation method: Most Frequent**

Although these variables contain ordered codes, their values represent discrete
categories or repayment-status levels. Most-frequent imputation is used to
avoid creating artificial category values that may not exist in the original
data.

### Leakage Prevention

All imputation statistics are learned **only from the corrupted training
data**.

The fitted imputers are then applied to the validation data without
re-estimating the statistics.

Therefore, information from the clean validation set is not used during the
imputation process.

The preprocessing is implemented inside a pipeline to ensure that the

## Missing Data Experiment — Results and Interpretation

The missing-data experiment evaluated Logistic Regression, Random Forest, and
Gradient Boosting under increasing levels of MCAR missingness: **0%, 5%, 10%,
20%, and 30%**.

The evaluation was performed on a clean validation set, while only the training
features were corrupted. Model configurations remained fixed across all
missingness levels.

### F1-score Analysis

The effect of missingness on F1-score was model-dependent.

**Logistic Regression** showed an overall decline in F1-score from 0.4623 at
0% missingness to 0.4498 at 30% missingness. The decline was not perfectly
monotonic, as a small recovery occurred at 30%.

**Random Forest** showed relatively stable F1-score across the tested
missingness levels. Its F1-score remained close to the clean-data baseline,
ranging from approximately 0.535 to 0.541.

**Gradient Boosting** exhibited non-monotonic F1 behavior. Its F1-score
increased at higher missingness levels and reached 0.4958 at 30% missingness,
compared with 0.4675 under clean training data. This result should not be
interpreted as evidence that missing data improves the model; repeated
corruption experiments would be required to determine whether the observed
pattern is stable.

### ROC-AUC Analysis

ROC-AUC showed a more consistent pattern across models.

All three models experienced a gradual decrease in ROC-AUC as missingness
increased from 0% to 30%.

- Logistic Regression: 0.7130 → 0.7071
- Random Forest: 0.7777 → 0.7715
- Gradient Boosting: 0.7776 → 0.7712

This indicates that increasing MCAR missingness was associated with a gradual
reduction in the models' class-ranking ability.

### Key Finding

The results suggest that model robustness to missing data depends on both the
model and the evaluation metric.

While ROC-AUC showed a consistent downward trend across all three models,
F1-score exhibited non-monotonic behavior, particularly for the ensemble
models.

Therefore, the hypothesis that increasing missingness would simply produce a
monotonic decrease in predictive performance is **not fully supported by the
F1-score results**. However, the consistent decline in ROC-AUC provides
evidence of a gradual reduction in predictive discrimination as training-data
missingness increases.

The results also demonstrate why multiple evaluation metrics are important
when studying robustness to imperfect data.

### Recall Behavior

An interesting observation is that recall does not decrease monotonically with
increasing missingness. In some models, recall increases as the level of
missingness increases.

This indicates that data corruption does not affect all evaluation metrics in
the same direction. Higher recall means that the model identifies a larger
proportion of the actual positive cases, but this may occur alongside a
reduction in precision.

Therefore, the increase in recall should not be interpreted as an improvement
in overall model robustness. Instead, it indicates a change in the model's
precision-recall trade-off under increasing missingness.

This observation further supports the use of multiple evaluation metrics when
studying robustness to imperfect data.

Real-world datasets may contain unusually large or small feature values due to
measurement errors, data-entry errors, unusual observations, or other data
collection issues.

Such extreme values can affect machine learning models differently. Some models
may be relatively robust to extreme observations, while others may be more
sensitive to them.

In this experiment, we systematically introduce synthetic outliers into the
training features and study how the performance of the selected machine
learning models changes.

### Experimental Setup

The following outlier levels will be investigated:

- **0%** — Clean training data
- **1%** — Low outlier contamination
- **5%** — Moderate outlier contamination

Outliers will be introduced only into the training feature matrix.

The validation set will remain completely clean and will be used to evaluate

### Hypothesis

**H2:** Increasing levels of feature outliers will negatively affect predictive

The experiment will also help determine whether some model types are more

### Outlier Definition

For this experiment, an outlier is defined as a feature value that lies
substantially beyond the typical range of that feature.

The Interquartile Range (IQR) method is used to characterize the distribution
of each numerical feature:

`
IQR = Q_3 - Q_1
`

The conventional IQR boundaries are:

`
Lower Bound = Q_1 - 1.5(IQR)
`

`
Upper Bound = Q_3 + 1.5(IQR)
`

These boundaries are used to understand the natural distribution of the
training features.

Importantly, naturally occurring extreme values in the original dataset are
not removed or modified. They may represent legitimate observations.

Instead, synthetic outliers are introduced into the training data in a
controlled manner. This allows the experiment to isolate the effect of
additional feature contamination on model robustness.

### Synthetic Outlier Generation

Synthetic outliers are introduced only into numerical training features.

For each selected numerical feature, the IQR-based upper and lower bounds are
calculated from the original training data.

For a selected cell, the original value is replaced by a value beyond one of
these bounds.

The direction of contamination is randomly chosen:

- **Upper-tail outlier:** a value is generated above the upper IQR boundary.
- **Lower-tail outlier:** a value is generated below the lower IQR boundary.

The magnitude of the injected outlier is controlled using the feature's IQR,
rather than using a fixed multiplier across all features.

This makes the corruption scale-aware because different numerical features
have very different ranges and distributions.

Only `X_train` is corrupted. The target variable, validation data, and test
data remain unchanged.

The same random seed is used for reproducibility.

```python
def introduce_outliers(X, outlier_rate, random_state=42):
    """
    Introduce synthetic outliers into numerical features.

    outlier_rate: proportion of numerical feature cells to corrupt.
    """

    X_corrupted = X.copy()

    rng = np.random.default_rng(random_state)

    # Select only numerical columns
    numerical_data = X_corrupted[numerical_features].copy()

    n_cells = numerical_data.shape[0] * numerical_data.shape[1]
    n_outliers = int(round(outlier_rate * n_cells))

    # Randomly select cells
    selected_indices = rng.choice(
        n_cells,
        size=n_outliers,
        replace=False
    )

    rows, cols = np.unravel_index(
        selected_indices,
        numerical_data.shape
    )

    for row, col in zip(rows, cols):

        feature = numerical_features[col]

        # Calculate IQR from the original training feature
        Q1 = X[feature].quantile(0.25)
        Q3 = X[feature].quantile(0.75)
        IQR = Q3 - Q1

        # Calculate IQR boundaries
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Randomly choose upper or lower contamination
        if rng.random() < 0.5:
            new_value = upper_bound + 3 * IQR
        else:
            new_value = lower_bound - 3 * IQR

        X_corrupted.iloc[row, X.columns.get_loc(feature)] = new_value

    return X_corrupted
```

The preprocessing strategy remains the same as that used during the clean

The evaluation procedure is:

1. Start with a specific level of feature outlier contamination.
2. Fit the preprocessing pipeline on the corrupted training data.
3. Train the frozen model configuration on the transformed training data.
4. Apply the fitted preprocessing pipeline to the **clean validation set**.
5. Generate predictions and predicted probabilities.
6. Evaluate the model using Accuracy, Precision, Recall, F1-score, and ROC-AUC.

The validation set is kept completely clean throughout the experiment. This

A fresh model instance is used for each outlier level so that models trained

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

def evaluate_model(model, X_train_corrupted, y_train, X_val, y_val):
    """
    Train a model on corrupted training data and evaluate
    it on the clean validation set.
    """

    # Train
    model.fit(X_train_corrupted, y_train)

    # Predictions
    y_pred = model.predict(X_val)
    y_prob = model.predict_proba(X_val)[:, 1]

    # Metrics
    results = {
        "Accuracy": accuracy_score(y_val, y_pred),
        "Precision": precision_score(y_val, y_pred),
        "Recall": recall_score(y_val, y_pred),
        "F1": f1_score(y_val, y_pred),
        "ROC-AUC": roc_auc_score(y_val, y_prob)
    }

    return results
```

## Experiment 3: Robustness to Label Noise

### Research Question

**RQ3:** How does increasing label noise in the training data affect model
performance and prediction reliability?

### Objective

In real-world machine learning datasets, labels may sometimes be incorrect due
to human annotation errors, data-entry mistakes, ambiguous cases, or imperfect
labeling processes.

A model trained on incorrect labels may learn misleading relationships between
the input features and the target variable.

In this experiment, we systematically introduce different levels of synthetic
label noise into the training data and study how the performance of the
selected machine learning models changes.

### Experimental Setup

The following label-noise levels will be investigated:

- **0%** — Clean training labels
- **5%** — Low label noise
- **10%** — Moderate label noise
- **20%** — High label noise
- **30%** — Severe label noise

Label noise will be introduced **only into the training labels**.

The feature data will remain unchanged, and the validation set will remain
completely clean.

Therefore:

- `X_train` → unchanged
- `y_train` → corrupted with label noise
- `X_val` → unchanged
- `y_val` → unchanged

The models will be trained using the corrupted training labels and evaluated
against the original clean validation labels.

### Experimental Principle

For each noise level:

1. Start with the original training labels.
2. Randomly select a specified proportion of training samples.
3. Flip their binary class labels.
4. Train the frozen model configurations using the corrupted labels.
5. Evaluate the models on the clean validation set.
6. Record Accuracy, Precision, Recall, F1-score, and ROC-AUC.

The same random seed and corruption procedure will be used across the
experiment to ensure reproducibility.

### Hypothesis

**H3:** Increasing levels of label noise will generally reduce predictive
performance and prediction reliability, although the magnitude of degradation
will differ across machine learning models.

Since label noise directly changes the target information available during
training, it is expected to have a stronger effect than some forms of feature
corruption.

### Label Noise Generation Method

Synthetic label noise is introduced by randomly flipping a specified proportion
of the binary training labels.

For a given noise level \(p\), approximately \(p\%\) of the training samples
are selected randomly.

For each selected sample, the original binary label is flipped:


`0 -> 1`


`
1 -> 0
`

The feature matrix remains unchanged. Only the training target variable is
corrupted.

The corruption is performed independently for each noise level using the
original clean training labels as the starting point. Therefore, noise levels
are not cumulative.

A fixed random seed is used to make the corruption process reproducible.

The validation labels are never modified and remain the clean reference for
evaluation.

```python
def introduce_label_noise(y, noise_rate, random_state=42):
    """
    Introduce synthetic label noise by randomly flipping
    a proportion of binary labels.

    noise_rate: proportion of labels to flip.
    """

    y_corrupted = y.copy()

    rng = np.random.default_rng(random_state)

    # Number of labels to corrupt
    n_samples = len(y_corrupted)
    n_noisy = int(round(noise_rate * n_samples))

    # Randomly select unique sample indices
    noisy_indices = rng.choice(
        n_samples,
        size=n_noisy,
        replace=False
    )

    # Flip binary labels: 0 -> 1 and 1 -> 0
    y_corrupted.iloc[noisy_indices] = (
        1 - y_corrupted.iloc[noisy_indices]
    )

    return y_corrupted
```

## Experiment 4: Model Disagreement

### Research Question

**RQ4:** Do different machine learning models disagree more frequently on samples that are difficult to classify or affected by data imperfections?

### Objective

The previous experiments examined how different forms of data corruption affect overall model performance. In this experiment, the analysis moves from dataset-level performance to **sample-level prediction behavior**.

Different machine learning models may learn different decision boundaries from the same training data. As a result, they may produce different predictions for individual samples.

The objective of this experiment is to determine whether such model disagreement is associated with difficult classification cases and prediction errors.

### Expanded Model Pool

The model pool is expanded from three to five models for this experiment:

* Logistic Regression
* Decision Tree
* Random Forest
* Gradient Boosting
* Support Vector Machine (SVM)

The additional models are introduced specifically to increase model diversity for the disagreement analysis. The robustness experiments conducted previously are not rerun.

### Experimental Setup

The five models will be trained using the clean training dataset and evaluated on the clean validation dataset.

For each validation sample, the predicted class from all five models will be recorded.

The predictions will then be compared to determine whether the models agree or disagree.

### Disagreement Measure

For each sample, the majority prediction among the five models will be identified.

The disagreement score is defined as:

$$
Disagreement =
1 -
\frac{\text{Number of models supporting the majority prediction}}
{\text{Total number of models}}
$$

With five models, the disagreement score can take the values:

* **0.0** — complete agreement
* **0.2** — one model disagrees with the other four
* **0.4** — two models disagree with the other three

A higher disagreement score therefore indicates greater divergence between the model predictions.

### Analysis

The disagreement scores will be compared with the true validation labels to investigate whether high-disagreement samples are more frequently misclassified.

Prediction confidence will also be examined to determine whether disagreement is associated with lower model confidence.

### Hypothesis

**H4:** Samples on which different machine learning models disagree more frequently are expected to be more difficult to classify and may have a higher probability of being misclassified.

The hypothesis will be evaluated empirically rather than assuming that disagreement necessarily indicates unreliability.


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
decision_tree_model = Pipeline(
    steps=[
        ("preprocessor", tree_preprocessor),
        ("model", DecisionTreeClassifier(
            max_depth=10,
            min_samples_leaf=5,
            class_weight="balanced",
            random_state=42
        ))
    ]
)
```

```python
svm_model = Pipeline(
    steps=[
        ("preprocessor", logistic_preprocessor),
        ("model", SVC(
            C=1.0,
            kernel="rbf",
            probability=True,
            class_weight="balanced",
            random_state=42
        ))
    ]
)
```

```python
disagreement_models = {
    "Logistic Regression": Pipeline(
        steps=[
            ("preprocessor", logistic_preprocessor),
            ("model", LogisticRegression(
                C=0.1,
                max_iter=1000,
                class_weight="balanced",
                random_state=42
            ))
        ]
    ),

    "Decision Tree": decision_tree_model,

    "Random Forest": Pipeline(
        steps=[
            ("preprocessor", tree_preprocessor),
            ("model", RandomForestClassifier(
                n_estimators=300,
                max_depth=10,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            ))
        ]
    ),

    "Gradient Boosting": Pipeline(
        steps=[
            ("preprocessor", tree_preprocessor),
            ("model", GradientBoostingClassifier(
                n_estimators=200,
                learning_rate=0.1,
                max_depth=3,
                random_state=42
            ))
        ]
    ),

    "SVM": svm_model
}
```

```python
model_columns = [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "Gradient Boosting",
    "SVM"
]

prediction_df["Majority_Prediction"] = (
    prediction_df[model_columns]
    .mode(axis=1)[0]
)

prediction_df.head()
```

```python
prediction_df["Disagreement"] = (
    1 -
    prediction_df[model_columns]
    .eq(prediction_df["Majority_Prediction"], axis=0)
    .mean(axis=1)
)
```

### Evaluating Majority Prediction Correctness

Prediction disagreement alone does not imply that a sample is unreliable.

To test **H4**, we compare the majority prediction of the five models with the true validation label.

For each validation sample:

- `Majority_Correct = True` if the majority prediction matches the true label.
- `Majority_Correct = False` otherwise.

We then calculate the error rate separately for each disagreement level.

If higher disagreement is associated with a higher error rate, this provides evidence that model disagreement is related to difficult classification cases.

```python
prediction_df["Majority_Correct"] = (
    prediction_df["Majority_Prediction"] == prediction_df["True_Label"]
)
```

```python
prediction_df.head()
```

```python
error_by_disagreement = (
    prediction_df.groupby("Disagreement")
    .agg(
        Samples=("Majority_Correct", "size"),
        Accuracy=("Majority_Correct", "mean")
    )
)

error_by_disagreement["Error_Rate"] = (
    1 - error_by_disagreement["Accuracy"]
)

error_by_disagreement
```

### Result and Interpretation

The error rate of the majority prediction increased consistently with model disagreement:

- **0.0 disagreement:** 14.61% error rate
- **0.2 disagreement:** 28.38% error rate
- **0.4 disagreement:** 39.16% error rate

This shows a clear association between model disagreement and prediction error. Samples on which more models disagreed were more frequently misclassified by the majority prediction.

The error rate increased by **24.55 percentage points** from complete agreement (0.0) to the highest observed disagreement level (0.4).

These results provide evidence supporting **H4**, suggesting that model disagreement can be a useful indicator of difficult-to-classify samples.

However, disagreement alone does not establish that a sample is unreliable. Therefore, prediction confidence will be examined next to determine whether high disagreement is also associated with lower model confidence.

### Prediction Confidence

Model disagreement indicates how much the five models differ in their predictions, but it does not tell us how confident the individual models are.

To complement the disagreement analysis, prediction probabilities are obtained from each model using `predict_proba()`.

For binary classification, the confidence of a model is defined as the larger of the predicted probabilities for the two classes:

`
Confidence = max(P(y=0), P(y=1))
`

A value closer to 1 indicates greater confidence, while a value closer to 0.5 indicates greater uncertainty.

We will examine whether samples with higher model disagreement also tend to have lower prediction confidence.

```python
probabilities = {}

for model_name, model in disagreement_models.items():
    probabilities[model_name] = model.predict_proba(X_val)[:, 1]
```

```python
probabilities
```

```python
confidence = {}

for model_name, probs in probabilities.items():
    confidence[model_name] = np.maximum(probs, 1 - probs)
```

```python
confidence
```

### Aggregating Model Confidence

Each model provides a confidence score for every validation sample.

Since the disagreement analysis considers the complete five-model pool, we summarize the individual model confidence scores using their mean.

The resulting **Mean Confidence** represents the average confidence of the five models for each sample.

A lower mean confidence indicates that the models are collectively more uncertain, while a higher value indicates stronger confidence across the model pool.

This measure will later be examined together with model disagreement to identify samples that may be difficult or potentially unreliable.

```python
confidence_df = pd.DataFrame(confidence, index=X_val.index)

confidence_df["Mean_Confidence"] = confidence_df.mean(axis=1)

confidence_df.head()
```

### Combining Disagreement and Confidence

The disagreement score captures how much the five models differ in their predictions, while mean confidence captures how certain the model pool is about the predictions.

To analyze these two signals together, the mean confidence score is added to the sample-level prediction table.

This allows us to examine whether samples with higher disagreement also tend to have lower confidence and higher prediction error.

```python
prediction_df["Mean_Confidence"] = confidence_df["Mean_Confidence"]
```

```python
prediction_df[
    ["Disagreement", "Mean_Confidence", "Majority_Correct"]
].head()
```

```python
confidence_by_disagreement = (
    prediction_df.groupby("Disagreement")
    .agg(
        Samples=("Mean_Confidence", "size"),
        Mean_Confidence=("Mean_Confidence", "mean"),
        Error_Rate=("Majority_Correct", lambda x: 1 - x.mean())
    )
)

confidence_by_disagreement
```

### Result and Interpretation

The relationship between model disagreement, mean confidence, and prediction error shows two consistent patterns.

As model disagreement increased from **0.0 to 0.4**, mean confidence decreased:

- **0.0 disagreement:** 0.7789 mean confidence
- **0.2 disagreement:** 0.7044 mean confidence
- **0.4 disagreement:** 0.6616 mean confidence

At the same time, the majority-prediction error rate increased:

- **0.0 disagreement:** 14.61%
- **0.2 disagreement:** 28.38%
- **0.4 disagreement:** 39.16%

Therefore, higher model disagreement was associated with both **lower prediction confidence** and **higher prediction error**.

These results provide stronger evidence supporting **H4**. Model disagreement appears to be a useful signal for identifying samples that are more difficult for the model pool to classify.

However, disagreement and low confidence alone do not establish that a sample is unreliable. The next stage will combine these signals to identify potentially unreliable or difficult-to-classify samples at the individual sample level.

## Unreliable Sample Detection

### Objective

The previous analysis showed that higher model disagreement is associated with lower confidence and higher prediction error.

We now investigate whether these signals can be combined to identify potentially unreliable or difficult-to-classify samples at the individual sample level.

A sample will be flagged as a **candidate unreliable sample** when:

- Model disagreement is at least **0.2**, meaning at least one of the five models disagrees with the majority prediction.
- Mean model confidence is below **0.70**, indicating relatively low confidence across the model pool.

The detection rule is defined before evaluating the results to avoid selecting thresholds based on the observed error rates.

The resulting flagged samples will then be compared with unflagged samples using their actual prediction error.

Importantly, a flagged sample is considered **potentially unreliable or difficult**, not definitively unreliable. The true validation label is used only to evaluate whether the detection rule is associated with prediction error.

```python
prediction_df["Potentially_Unreliable"] = (
    (prediction_df["Disagreement"] >= 0.2) &
    (prediction_df["Mean_Confidence"] < 0.70)
)

prediction_df["Potentially_Unreliable"].value_counts()
```

### Evaluating the Detection Rule

The detection rule identified 355 validation samples as potentially unreliable, while 4,145 samples were not flagged.

To evaluate whether the rule is useful, we compare the prediction error rate between the two groups.

If the flagged group has a substantially higher error rate than the unflagged group, this would provide evidence that the combination of model disagreement and low confidence can help identify difficult-to-classify samples.

The evaluation uses the true validation labels only after the detection rule has been defined.

```python
reliability_evaluation = (
    prediction_df.groupby("Potentially_Unreliable")
    .agg(
        Samples=("Majority_Correct", "size"),
        Accuracy=("Majority_Correct", "mean")
    )
)

reliability_evaluation["Error_Rate"] = (
    1 - reliability_evaluation["Accuracy"]
)

reliability_evaluation
```

### Result and Interpretation

The detection rule flagged **355 out of 4,500 validation samples** as potentially unreliable.

The error rates of the two groups were:

- **Not flagged:** 18.96% error rate
- **Potentially unreliable:** 43.10% error rate

The flagged group therefore had an error rate approximately **24.14 percentage points higher** than the unflagged group.

This provides evidence that combining **higher model disagreement** with **lower mean prediction confidence** can identify a subset of samples that are substantially more difficult for the model pool to classify.

However, the flagged samples should not be interpreted as definitively unreliable because no independent ground-truth reliability label is available. Instead, the detection rule should be interpreted as a **risk indicator for potentially difficult or unreliable predictions**.

Overall, the result supports **H5**, indicating that model disagreement combined with prediction confidence can be useful for identifying potentially problematic samples.

```python
error_detection = pd.crosstab(
    prediction_df["Potentially_Unreliable"],
    prediction_df["Majority_Correct"],
    rownames=["Potentially_Unreliable"],
    colnames=["Majority_Correct"]
)

error_detection
```

### Error-Capture Analysis

The detection rule flagged 355 out of 4,500 validation samples, representing **7.89%** of the validation set.

Among all validation samples, there were **939 majority-prediction errors**. The detection rule identified **153** of these errors.

Therefore, the error-capture rate was:

`
153/939 x 100 = 16.30%
`

Although the detector captured only a subset of all errors, it concentrated a relatively large number of errors within a small subset of samples.

However, 202 of the 355 flagged samples were correctly classified by the majority prediction. This indicates that the detection rule produces false alarms and should therefore be interpreted as a **risk-ranking or screening mechanism**, rather than a definitive error detector.

Overall, the analysis suggests that combining model disagreement and prediction confidence can identify a subset of samples with substantially elevated prediction risk, while also highlighting the limitations of using these signals alone.

# Cross-Experiment Analysis

The previous experiments investigated different forms of data imperfection independently. While each experiment provides evidence about a particular type of corruption, examining the results separately does not fully reveal the broader behavior of machine learning models under imperfect data.

The objective is not simply to identify which experiment produced the largest numerical change. Instead, the goal is to determine **what the combined evidence tells us about the reliability and robustness of machine learning systems when the training data is imperfect.**

## 1. Objectives of the Cross-Experiment Analysis

The cross-experiment analysis addresses the following questions:

### Q1. How do different types of data imperfection affect model performance?

The study introduced three different forms of synthetic data corruption:

* **Missing data** — random feature values were removed from the training set.
* **Feature outliers** — extreme numerical feature values were introduced.
* **Label noise** — training labels were randomly flipped.

We compare the resulting performance changes to determine whether different forms of imperfection produce similar or different effects.

---

### Q2. Are some machine learning models more sensitive to particular types of corruption?

The experiments used multiple model families, including:

* Logistic Regression
* Random Forest
* Gradient Boosting

These models make different assumptions and learn different types of decision boundaries.

Therefore, the same data imperfection may affect them differently.

The analysis will examine whether the observed performance degradation is **model-dependent rather than uniform across all models**.

---

### Q3. Which evaluation metrics provide the clearest picture of robustness?

Because the target variable is imbalanced, accuracy alone is not sufficient to characterize model behavior.

Therefore, the cross-experiment analysis considers:

* Accuracy
* Precision
* Recall
* F1-score
* ROC-AUC

Particular attention is given to **F1-score and ROC-AUC**, since they provide more informative views of classification performance under the observed class imbalance.

The analysis will also distinguish between changes in different metrics rather than treating a single metric as a complete representation of model performance.

---

## 2. Experimental Comparison Framework

To make the experiments comparable, the analysis considers the following common structure:

$$
\text{Clean Training Data}
\rightarrow
\text{Controlled Data Corruption}
\rightarrow
\text{Model Training}
\rightarrow
\text{Evaluation on Clean Validation Data}
$$

The corruption was applied only to the training data in the robustness experiments, while the validation set remained clean.

This design allows the analysis to measure how imperfections in the **training data** affect the resulting model's ability to generalize to previously unseen, clean data.

This distinction is important because the experiments are intended to study **robustness to imperfect training data**, rather than performance when both training and evaluation data are corrupted.

---

## 3. Types of Imperfection Studied

The three corruption mechanisms represent different failure modes in a machine learning pipeline.

### Missing Data

Missing-data corruption removes observed feature values while preserving the remaining information.

This represents situations in which:

* measurements are unavailable,
* records are incomplete,
* sensors fail,
* users do not provide information, or
* data collection systems contain missing entries.

The experiment investigated missingness levels from **0% to 30%**.

---

### Feature Outliers

Outlier corruption introduces extreme values into numerical feature cells.

This represents situations such as:

* measurement errors,
* incorrect numerical entries,
* sensor anomalies,
* unusual observations, or
* data-entry errors.

The experiment investigated outlier rates from **0% to 10%** of numerical feature cells.

Importantly, the experiment did not assume that naturally occurring extreme observations in the original dataset were erroneous. The corruption mechanism deliberately introduced additional synthetic extremes so that the effect could be studied in a controlled manner.

---

### Label Noise

Label-noise corruption changes the target label associated with randomly selected training observations.

This represents situations in which:

* observations are incorrectly annotated,
* human labeling errors occur,
* classification rules are applied inconsistently, or
* target information is recorded incorrectly.

The experiment investigated label-noise levels from **0% to 30%**.

Unlike feature corruption, label noise directly changes the supervision provided to the learning algorithm.

---

## 4. Why Direct Comparison Requires Caution

Although the three experiments are compared in this section, their corruption rates should **not be interpreted as directly equivalent quantities**.

For example:

* 20% missingness means approximately 20% of feature cells were made missing.
* 10% outlier corruption means approximately 10% of numerical feature cells were replaced with synthetic extreme values.
* 20% label noise means approximately 20% of training labels were flipped.

These represent different units of corruption and different mechanisms of information loss.

Therefore, the analysis will focus primarily on **patterns of performance degradation and model sensitivity**, rather than claiming that one corruption level is inherently "worse" than another simply because its percentage is larger.

---

# 5. Model Robustness Perspective

A central objective of this study is to determine whether model robustness is a universal property of a model or whether it depends on the type of imperfection present in the data.

A model may remain relatively stable under one type of corruption while being more sensitive to another.

Therefore, the analysis will examine robustness as a **relationship between three factors**:

$$
\text{Model}
\quad \times \quad
\text{Data Imperfection}
\quad \times \quad
\text{Performance Metric}
$$

This provides a more meaningful view of robustness than assigning a single overall robustness score to each model.

---

# 6. Performance Degradation

For each corruption type, the clean-data performance provides the reference point.

The change in a metric can be expressed as:

$$
\Delta M = M_{\text{corrupted}} - M_{\text{clean}}
$$

where \(M\) represents a performance metric such as F1-score or ROC-AUC.

A negative value indicates a reduction in performance after corruption.

For example, if ROC-AUC changes from 0.78 to 0.75:

$$
\Delta ROC\text{-}AUC = 0.75 - 0.78 = -0.03
$$

This allows performance changes to be interpreted consistently across the different corruption experiments.

However, because the corruption mechanisms operate on different units and scales, these changes will be used to study **sensitivity patterns**, not to construct an unsupported universal ranking of corruption severity.

---

# 7. Model-Specific Sensitivity

The analysis will examine whether the three primary models respond differently to the same corruption.

This is particularly important because the models represent different learning mechanisms:

* **Logistic Regression** learns a linear decision boundary.
* **Random Forest** combines multiple decision trees and can model nonlinear relationships.
* **Gradient Boosting** builds an ensemble sequentially, with later trees focusing on previous errors.

Differences in their responses can therefore provide evidence that robustness is influenced by the interaction between the **learning algorithm and the type of data imperfection**.

The analysis will avoid interpreting one model as universally superior. Instead, it will describe the specific performance patterns observed under each experimental condition.

---

# 8. Reliability Beyond Aggregate Performance

The first three experiments primarily evaluate robustness using dataset-level performance metrics.

Experiment 4 extends the analysis to the **individual prediction level**.

Five models were used to examine:

* prediction agreement,
* prediction disagreement,
* prediction confidence,
* majority-prediction error, and
* potentially unreliable samples.

This introduces an important distinction:

> A model can have acceptable overall performance while still producing individual predictions for which its reliability is uncertain.

Therefore, aggregate metrics such as F1 and ROC-AUC are complemented by sample-level reliability analysis.

---

# 9. Model Disagreement as a Reliability Signal

Experiment 4 showed that samples with greater disagreement among the five models had higher majority-prediction error rates.

The observed error rates were:

* **0.0 disagreement:** 14.61%

### Interpretation principle

The direction of change in a metric must be interpreted according to what the metric measures.

For example:

* A decrease in **precision** indicates that a larger proportion of predicted positive cases are incorrect.
* A decrease in **recall** indicates that the model is missing more actual positive cases.
* A decrease in **F1-score** indicates deterioration in the balance between precision and recall.
* A decrease in **ROC-AUC** indicates reduced class-discrimination ability.
* Changes in **accuracy** must be interpreted carefully because the dataset has an imbalanced target distribution.

The analysis will therefore examine the metrics individually rather than assuming that all metrics should move together.

### Important distinction

A non-monotonic metric trajectory does not automatically imply that the experiment failed.

It may indicate that the corruption mechanism affects different aspects of the classifier differently.

Such behavior is itself an experimental observation and will be interpreted rather than removed or smoothed.

The ROC-AUC values were 0.7182 for Logistic Regression, 0.7686 for Random Forest, and 0.7811 for Gradient Boosting, indicating differences in their ability to distinguish between the two classes.

Overall, the test-set evaluation confirms that the models exhibit different trade-offs between Precision and Recall. This reinforces the importance of evaluating multiple metrics rather than relying on Accuracy alone, particularly for the imbalanced target distribution in this dataset.

The test set was used only for this final evaluation and was not used to modify model configurations or experimental decisions.


# Final Research Findings and Discussion

The experiments investigated the robustness and reliability of machine-learning models under three forms of data imperfection: missing feature values, feature outliers, and label noise. A separate analysis examined whether model disagreement and prediction confidence could provide sample-level signals of difficult or potentially unreliable predictions.

The findings below address the five research questions defined at the beginning of the study.

---

## RQ1 — Effect of Missing Data

**Research Question:**
How does increasing missingness affect predictive performance and reliability of different ML models?

Increasing missingness generally caused a gradual decline in several performance measures, particularly **ROC-AUC, Accuracy, and Precision**.

The decline in ROC-AUC was observed across all three models as missingness increased from 0% to 30%. This indicates that increasing missing feature values can gradually reduce the models' ability to distinguish between the two target classes.

However, the effect was not uniform across all metrics. Recall and F1-score showed model-dependent and non-monotonic behavior. Random Forest remained relatively stable in F1-score, while Gradient Boosting showed an increase in F1-score at higher missingness levels.

Therefore, the results provide **partial support for H1**. Missing data was associated with degradation in several performance measures, but the effect was not uniformly negative across every metric.

---

## RQ2 — Effect of Feature Outliers

**Research Question:**
How sensitive are different ML models to increasing levels of feature outliers?

The synthetic feature-outlier experiment produced relatively small and non-monotonic changes in the evaluation metrics.

Random Forest remained comparatively stable across the tested outlier levels. Logistic Regression showed a small decline in ROC-AUC but slight increases in Accuracy and F1-score. Gradient Boosting also showed only modest changes.

Therefore, the experiment does not provide evidence of a uniform performance degradation as the outlier rate increased.

The findings provide **partial support for H2** because some model- and metric-specific sensitivity was observed, but the effect was limited under the particular synthetic outlier mechanism used in this study.

These results should not be interpreted as evidence that real-world outliers are harmless. Different outlier structures, magnitudes, or correlated feature contamination could produce different effects.

---

## RQ3 — Effect of Label Noise

**Research Question:**
How does increasing label noise affect model performance and prediction reliability?

Label noise produced the clearest performance degradation among the three corruption experiments.

As random label noise increased from 0% to 30%, Accuracy, Precision, F1-score, and ROC-AUC generally decreased across the models.

The effect was particularly visible for Logistic Regression in Accuracy and for Gradient Boosting in ROC-AUC.

Recall behaved differently, especially for Logistic Regression, where Recall increased as label noise increased. This demonstrates that label corruption can change the prediction behavior of a model rather than simply reducing every metric simultaneously.

Overall, the findings provide **support for H3**. Increasing random label noise generally reduced predictive performance and class-separation ability.

---

## RQ4 — Model Disagreement and Difficult Samples

**Research Question:**
Do different ML models disagree more frequently on samples that are difficult to classify or affected by data imperfections?

The five-model disagreement analysis showed a clear relationship between model disagreement and prediction error.

When all five models agreed, the majority-prediction error rate was **14.61%**.

When one model disagreed with the majority, corresponding to a disagreement score of 0.2, the error rate increased to **28.38%**.

When two models disagreed with the majority, corresponding to a disagreement score of 0.4, the error rate increased further to **39.16%**.

Thus, the error rate increased by approximately **24.55 percentage points** between complete agreement and the highest observed disagreement level.

This provides evidence supporting **H4**. Higher model disagreement was associated with a higher frequency of prediction errors and therefore appears to be a useful signal of difficult-to-classify samples.

However, disagreement itself does not prove that a sample is unreliable. It is better interpreted as a signal of prediction difficulty or model uncertainty.

---

## RQ5 — Detecting Potentially Unreliable Samples

**Research Question:**
Can model disagreement and prediction confidence be used to identify potentially unreliable or difficult-to-classify samples?

The analysis combined two signals:

* disagreement score ≥ 0.2
* mean model confidence < 0.70

Using this rule, **355 of 4,500 validation samples (7.89%)** were flagged as potentially unreliable.

The flagged samples had an error rate of **43.10%**, compared with **18.96%** for the unflagged samples.

Therefore, the flagged subset had an error rate approximately **24.14 percentage points higher** than the unflagged subset.

The detection rule also captured **153 of the 939 majority-prediction errors**, corresponding to an error-capture rate of **16.30%**.

This result provides **partial support for H5**.

The method successfully concentrated prediction errors within a relatively small subset of samples, but it did not identify all errors. Furthermore, 202 of the 355 flagged samples were correctly classified.

Therefore, the proposed approach should be viewed as a **risk-screening mechanism rather than a definitive unreliable-sample detector**.

---

# Overall Research Findings

The combined experiments lead to several broader observations.

### 1. Robustness depends on the type of data imperfection

The models did not respond identically to missing data, feature outliers, and label noise.

Label noise produced the clearest degradation across several metrics, while missingness produced more gradual changes. Feature outliers produced comparatively limited effects under the specific synthetic corruption mechanism used.

This demonstrates that robustness cannot be characterized using a single corruption type.

---

### 2. Model robustness is metric-dependent

A model can remain relatively stable in one metric while changing substantially in another.

For example, Recall sometimes increased while Precision decreased under corruption. Similarly, F1-score did not always follow the same trend as ROC-AUC.

Therefore, evaluating robustness using only Accuracy can hide important changes in model behavior.

---

### 3. Different models respond differently to the same corruption

Logistic Regression, Random Forest, and Gradient Boosting showed different response patterns under the same corruption levels.

This suggests that model sensitivity is influenced by both the underlying learning algorithm and the type of imperfection introduced into the data.

The experiments therefore do not establish a universal ranking of model robustness.

Instead, they demonstrate that robustness is **corruption-specific and model-dependent**.

---

### 4. Aggregate performance is not sufficient to understand reliability

The first three experiments focused primarily on dataset-level performance.

Experiment 4 extended the analysis to individual predictions.

The disagreement analysis showed that samples with greater disagreement among models had higher error rates and lower mean confidence.

This demonstrates that aggregate metrics alone may not reveal which individual samples are responsible for prediction uncertainty.

---

### 5. Model disagreement provides a useful sample-level signal

The five-model analysis showed:

* Complete agreement → 14.61% error
* 0.2 disagreement → 28.38% error
* 0.4 disagreement → 39.16% error

The increasing error rate suggests that model disagreement contains useful information about prediction difficulty.

However, disagreement should be interpreted as a **risk signal**, not as proof that a prediction is incorrect.

---

### 6. Combining disagreement and confidence improves risk screening

The combination of disagreement and confidence identified a relatively small subset of samples with substantially higher prediction error.

Only **7.89%** of validation samples were flagged, but this group had a **43.10% error rate**, compared with **18.96%** among unflagged samples.

This suggests that disagreement and confidence can potentially be used to prioritize samples for:

* human review,
* additional data collection,
* label verification,
* model inspection,
* or further investigation.

The current experiment demonstrates the usefulness of the signal, but does not establish that the method is universally effective.

---

# Answer to the Central Research Problem

The central objective of this study was to investigate whether machine-learning reliability can be better understood by studying both **data imperfections** and **sample-level prediction behavior**.

The experiments indicate that:

> **Data imperfections can affect machine-learning models in different ways depending on the corruption type, model, and evaluation metric. Furthermore, model disagreement combined with prediction confidence can provide a useful signal for identifying samples with elevated prediction risk.**

The results therefore support a broader view of machine-learning robustness in which reliability is not represented by a single performance score.

Instead, reliability should be examined across multiple dimensions:

**Data Quality → Model Robustness → Prediction Behavior → Sample-Level Risk**

# Limitations and Future Work

Although the experiments provide useful evidence about machine-learning robustness and prediction reliability, several limitations should be considered when interpreting the findings.

## Limitations

### 1. Single Dataset

The experiments were conducted on the **UCI Default of Credit Card Clients** dataset.

Therefore, the observed robustness patterns may depend on the characteristics of this particular dataset, including its feature distributions, class imbalance, and relationship between repayment behavior and default.

The findings cannot automatically be generalized to all datasets or application domains.

---

### 2. Synthetic Data Corruption

Missing values, feature outliers, and label noise were introduced artificially.

The experiments therefore represent controlled corruption mechanisms rather than every form of real-world data imperfection.

For example:

* Missing values were generated using an MCAR-style random cell-level mechanism.
* Feature outliers were generated using an IQR-based procedure.
* Label noise was introduced through random label flipping.

Real-world corruption may be structured, systematic, or concentrated within particular groups of observations.

---

### 3. Limited Number of Random Seeds

The main experiments were initially conducted using a fixed random seed.

A smaller multi-seed reproducibility analysis was subsequently performed for selected corruption levels.

The reproducibility results showed that the general patterns were broadly stable, but some numerical variation remained across seeds.

A more comprehensive study would evaluate every corruption level across multiple random seeds and report distributions or confidence intervals rather than relying primarily on single-run results.

---

### 4. Limited Model Pool

The robustness experiments focused on three core models:

* Logistic Regression
* Random Forest
* Gradient Boosting

The disagreement experiment expanded the model pool to five models by adding Decision Tree and SVM.

Other model families, such as neural networks, XGBoost, LightGBM, or other ensemble methods, were not systematically evaluated.

Consequently, the observed model disagreement patterns may change when a substantially different model pool is used.

---

### 5. Fixed Corruption Mechanisms

Each corruption experiment used one predefined mechanism.

For example, the outlier experiment modified individual numerical cells independently.

In real datasets, outliers may affect several correlated features simultaneously, and missingness may follow mechanisms such as MAR or MNAR rather than MCAR.

Future experiments should therefore evaluate multiple corruption mechanisms.

---

### 6. Threshold Selection for Unreliable-Sample Detection

The potentially unreliable sample detector used the predefined rule:

* disagreement ≥ 0.2
* mean confidence < 0.70

These thresholds were selected before evaluating the detection results, which avoids directly optimizing them on the observed validation errors.

However, the thresholds themselves are still heuristic.

A stronger study could evaluate threshold sensitivity systematically or learn the thresholds using a separate validation procedure.
